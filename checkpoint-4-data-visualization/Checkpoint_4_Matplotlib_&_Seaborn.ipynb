### Outliers values

Pada tahap ini, kita akan mendeteksi nilai ekstrem (outliers) pada kolom-kolom numerik
menggunakan metode **IQR (Interquartile Range)**. Nilai dianggap outlier jika berada
di luar rentang: Q1 - 1.5×IQR hingga Q3 + 1.5×IQR.

Perlu diingat bahwa tidak semua outlier harus dihapus — kita perlu memahami konteks
kolom tersebut terlebih dahulu sebelum memutuskan penanganannya.

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Berdasarkan hasil deteksi outliers di atas, terdapat **3 kolom** yang mengandung outliers:

* Kolom **`quality_of_faculty`** → Persentase outlier tertinggi sebesar **±19,27%**
(sekitar 424 baris). Ini menunjukkan bahwa terdapat banyak universitas yang memiliki
nilai ranking kualitas fakultas yang sangat menyimpang dari mayoritas, kemungkinan
karena hanya sebagian universitas yang benar-benar unggul secara signifikan dalam
kualitas tenaga pengajarnya.

* Kolom **`score`** → Persentase outlier sebesar **±11,00%** (sekitar 242 baris).
Outlier pada kolom skor total ini mencerminkan adanya kesenjangan yang cukup besar
antara universitas-universitas papan atas dunia (skor mendekati 100) dengan universitas
pada peringkat bawah yang skornya jauh lebih rendah.

* Kolom **`national_rank`** → Persentase outlier sebesar **±10,55%** (sekitar 232 baris).
Outlier di sini wajar terjadi karena negara seperti USA memiliki jumlah universitas yang
masuk peringkat jauh lebih banyak dibanding negara lain, sehingga nilai `national_rank`

* Kolom **`year`** → Persentase outlier sebesar **±4,55%** (100 baris). Hal ini
disebabkan oleh data tahun 2012 dan 2013 yang hanya memiliki 100 baris masing-masing,
jauh lebih sedikit dibanding 2014–2015 (1.000 baris), sehingga secara statistik

**Kesimpulan:** Outlier yang ditemukan pada dataset ini sebagian besar merupakan

## **Composition/Komposisi**

Aktivitas: Melihat proporsi atau komposisi jumlah universitas berdasarkan negara asal
untuk 5 negara teratas.

Tujuan: Mengetahui seberapa besar kontribusi masing-masing negara dalam membentuk

masing-masing berkontribusi relatif kecil dibandingkan USA.
* Kategori **Others** (negara-negara selain top 5) masih mendominasi sisa **47,4%**,

**Insight:** Meskipun ada 59 negara dalam dataset, lebih dari **52%** universitas

Dengan demikian, tidak diperlukan penanganan khusus untuk duplicated values pada dataset ini.

### Outliers Values

```python
print("Jumlah Outlier per Kolom")

outlier_summary = {}
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    outlier_summary[col] = len(outliers)
    print(f"{col}: {len(outliers)} outlier | lower={lower:.2f}, upper={upper:.2f}")
```

Berdasarkan hasil deteksi outlier menggunakan metode IQR, ditemukan outlier pada kolom `score` sebanyak 242 data dan `quality_of_faculty` sebanyak 424 data. Namun setelah ditelaah lebih lanjut, nilai-nilai tersebut bukan merupakan kesalahan input melainkan data yang valid secara kontekstual. Kolom `score` memang memiliki rentang nilai yang lebar karena mencerminkan perbedaan performa nyata antara universitas peringkat atas dan bawah. Begitu pula `quality_of_faculty` yang merupakan kolom ranking dengan rentang alami dari 1 hingga ratusan. Oleh karena itu, tidak dilakukan penanganan khusus terhadap outlier pada dataset ini agar informasi asli dari data tetap terjaga.

## Data Construction

```python
df['academic_score'] = df[['publications', 'influence', 'citations']].mean(axis=1).round(2)
```

```python
def kategorikan_rank(rank):
    if rank <= 100:
        return 'Top 100'
    elif rank <= 500:
        return 'Top 500'
    else:
        return 'Top 1000'

df['rank_category'] = df['world_rank'].apply(kategorikan_rank)
```

```python
def kategorikan_score(score):
    if score >= 80:
        return 'Excellent'
    elif score >= 60:
        return 'Good'
    elif score >= 40:
        return 'Average'
    else:
        return 'Below Average'

df['score_category'] = df['score'].apply(kategorikan_score)
```

```python
print(df[['world_rank', 'academic_score', 'rank_category', 'score_category']].head(10))
```

Konstruksi data dilakukan dengan menambahkan tiga kolom baru yang relevan dengan studi kasus. Pertama, kolom `academic_score` dibuat dengan menghitung rata-rata dari kolom `publications`,` influence,` dan `citations` sebagai representasi performa akademik suatu universitas secara keseluruhan. Kedua, kolom `rank_category` dibuat untuk mengelompokkan universitas ke dalam tiga kategori berdasarkan `world_rank`, yaitu Top 100, Top 500, dan Top 1000, guna mempermudah analisis komparatif antar kelompok peringkat. Ketiga, kolom `score_category` dibuat untuk mengklasifikasikan universitas berdasarkan nilai `score` ke dalam kategori Excellent, Good, Average, dan Below Average, sehingga performa universitas dapat dianalisis secara lebih intuitif dan deskriptif.

Aktivitas: Menganalisis kekuatan dan arah hubungan antar seluruh
metrik penilaian numerik dalam dataset.

Tujuan: Mengetahui pasangan variabel mana yang saling berkorelasi
kuat, sehingga dapat dipahami dinamika penilaian peringkat universitas
secara menyeluruh.

Visualisasi: Heatmap Korelasi.

```python
num_cols = ['score', 'quality_of_education', 'alumni_employment',
            'quality_of_faculty', 'publications', 'influence',
            'citations', 'patents', 'academic_score']

plt.figure(figsize=(11, 9))
mask = pd.DataFrame(False, index=num_cols, columns=num_cols)
# Upper triangle mask
import numpy as np
mask_arr = np.triu(np.ones_like(df[num_cols].corr(), dtype=bool), k=1)
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm',
            fmt='.2f', linewidths=0.5, square=True)
plt.title('Heatmap Korelasi Antar Metrik Penilaian Universitas',
          fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

* **`publications`, `influence`, `citations`** saling berkorelasi

* **`score`** berkorelasi negatif dengan semua metrik ranking
ini bersifat logis karena metrik ranking menggunakan skala
**angka kecil peringkat lebih baik**.

**Insight & Tindakan:** Untuk meningkatkan skor universitas secara
