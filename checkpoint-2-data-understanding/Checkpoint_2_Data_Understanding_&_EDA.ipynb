### Outliers values

Pada tahap ini, kita akan mendeteksi nilai ekstrem (outliers) pada kolom-kolom numerik
menggunakan metode **IQR (Interquartile Range)**. Nilai dianggap outlier jika berada
di luar rentang: Q1 - 1.5×IQR hingga Q3 + 1.5×IQR.

Perlu diingat bahwa tidak semua outlier harus dihapus — kita perlu memahami konteks
kolom tersebut terlebih dahulu sebelum memutuskan penanganannya.

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Berdasarkan hasil deteksi outliers di atas, terdapat **4 kolom** yang mengandung outliers:

* Kolom **`quality_of_faculty`** → Persentase outlier tertinggi sebesar **±19,27%**
(sekitar 424 baris). Ini menunjukkan bahwa terdapat banyak universitas yang memiliki
nilai ranking kualitas fakultas yang sangat menyimpang dari mayoritas, kemungkinan
karena hanya sebagian universitas yang benar-benar unggul secara signifikan dalam
kualitas tenaga pengajarnya.

* Kolom **`score`** → Persentase outlier sebesar **±11,00%** (sekitar 242 baris).
Outlier pada kolom skor total ini mencerminkan adanya kesenjangan yang cukup besar
antara universitas-universitas papan atas dunia (skor mendekati 100) dengan universitas
pada peringkat bawah yang skornya jauh lebih rendah.

* Kolom **`national_rank`** → Persentase outlier sebesar **±10,55%** (sekitar 232 baris).
Outlier di sini wajar terjadi karena negara seperti USA memiliki jumlah universitas yang
masuk peringkat jauh lebih banyak dibanding negara lain, sehingga nilai `national_rank`

* Kolom **`year`** → Persentase outlier sebesar **±4,55%** (100 baris). Hal ini
disebabkan oleh data tahun 2012 dan 2013 yang hanya memiliki 100 baris masing-masing,
jauh lebih sedikit dibanding 2014–2015 (1.000 baris), sehingga secara statistik

## **Composition/Komposisi**

Aktivitas: Melihat proporsi atau komposisi jumlah universitas berdasarkan negara asal
untuk 5 negara teratas.

Tujuan: Mengetahui seberapa besar kontribusi masing-masing negara dalam membentuk

masing-masing berkontribusi relatif kecil dibandingkan USA.
* Kategori **Others** (negara-negara selain top 5) masih mendominasi sisa **47,4%**,

**Insight:** Meskipun ada 59 negara dalam dataset, lebih dari **52%** universitas

## **Distribution/Distribusi**

Aktivitas: Menganalisis penyebaran nilai skor keseluruhan (score) universitas
dalam dataset.

Tujuan: Melihat apakah mayoritas universitas memiliki skor yang rendah atau tinggi,
serta mendeteksi adanya kecondongan (skewness) dalam distribusi skor.

Visualisasi: Histogram dengan KDE.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['score'], bins=30, kde=True, color='steelblue')
plt.title('Distribusi Skor Keseluruhan Universitas (Score)')
plt.xlabel('Score')
plt.ylabel('Frekuensi')
plt.show()
```

Berdasarkan visualisasi distribusi di atas:

* Distribusi `score` sangat **right-skewed (condong ke kanan)** — mayoritas
universitas memiliki skor yang **terkonsentrasi di kisaran 43–50** (rentang rendah).
* Hanya sebagian kecil universitas yang memiliki skor tinggi (di atas 60–100),
yang merupakan universitas-universitas papan atas seperti Harvard, MIT, dan Stanford.
* Nilai **mean (47,80)** jauh lebih besar dari **median (45,10)**, yang
mengkonfirmasi adanya right-skew akibat tarikan nilai ekstrem di ujung kanan.

**Insight:** Terdapat jurang yang sangat besar antara universitas-universitas
elite (skor mendekati 100) dengan universitas kebanyakan (skor 43–50).
Ini menggambarkan bahwa kualitas universitas kelas dunia tidak terdistribusi
secara merata — hanya segelintir institusi yang benar-benar unggul secara
signifikan dibanding ribuan universitas lainnya.

## **Relationship/Hubungan**

Aktivitas: Menganalisis korelasi antara skor keseluruhan (score) dengan
metrik-metrik penilaian lainnya seperti kualitas pendidikan, kualitas
fakultas, publikasi, dan lain-lain.

Tujuan: Mengetahui faktor/metrik penilaian mana yang paling kuat
hubungannya dengan skor akhir universitas.

Visualisasi: Heatmap Korelasi.

```python
num_cols = ['score', 'quality_of_education', 'alumni_employment',
            'quality_of_faculty', 'publications', 'influence',
            'citations', 'patents']

plt.figure(figsize=(10, 8))
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heatmap Korelasi Antar Metrik Penilaian Universitas')
plt.show()
```

Berdasarkan heatmap korelasi di atas, dapat dianalisis hubungan antar
metrik penilaian universitas:

* **`quality_of_faculty`** memiliki korelasi **negatif terkuat** terhadap
`score` sebesar **-0,69** — semakin kecil nilai ranking kualitas fakultas
(artinya semakin baik peringkatnya), semakin tinggi skor universitas tersebut.
* **`quality_of_education` (-0,60)**, **`influence` (-0,52)**,
**`publications` (-0,52)**, dan **`citations` (-0,52)** juga memiliki
korelasi negatif yang kuat terhadap `score`.
* Korelasi **negatif** di sini bersifat logis karena kolom-kolom tersebut
adalah **nilai ranking** (angka kecil = peringkat lebih baik), sehingga
universitas dengan peringkat metrik yang baik (angka kecil) akan memiliki
skor akhir yang tinggi.

**Insight:** Kualitas fakultas (`quality_of_faculty`) adalah **faktor
paling dominan** yang menentukan skor akhir sebuah universitas dalam
peringkat ini. Artinya, universitas yang ingin meningkatkan peringkat
globalnya perlu memprioritaskan peningkatan kualitas tenaga pengajar
dan akademisinya di atas faktor-faktor lainnya.
